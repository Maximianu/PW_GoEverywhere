```python
import re
from pathlib import Path
root = Path('src')
px_pattern = re.compile(r'(?P<num>\d+(?:\.\d+)?)px')
updated = []
for path in sorted(root.rglob('*')):
    if path.suffix not in {'.css', '.vue'}:
        continue
    text = path.read_text(encoding='utf-8')
    def repl(m):
        n = float(m.group('num'))
        if n == 0:
            return '0'
        rem = round(n / 16, 4)
        if rem.is_integer():
            return f'{int(rem)}rem'
        return f'{rem:.4f}'.rstrip('0').rstrip('.') + 'rem'
    new_text = px_pattern.sub(repl, text)
    if new_text != text:
        path.write_text(new_text, encoding='utf-8')
        updated.append(str(path))
print('Updated files:')
for file in updated:
    print(file)
```
